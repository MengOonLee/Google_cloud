# Cloud Speech API: 3 Ways

## Task 1. Create an API key

```python
1. To create an API key, on the Navigation menu click APIs & services > Credentials.

2. Click Create credentials and select API key.

3. Copy and record the key you just generated to use later in this lab.

4. Click Close.

5. On the Navigation menu click Compute Engine > VM Instances.

6. Click SSH to the right of the linux-instance VM name.

7. In the shell (SSH) run the following,
export API_KEY="API"
```

## Task 2. Create synthetic speech from text using the Text-to-Speech API

```python
1. source venv/bin/activate

2. nano synthesize-text.json

3. curl -s -X POST -H "Content-Type: application/json; charset=utf-8" \
  -d @synthesize-text.json \
  "https://texttospeech.googleapis.com/v1/text:synthesize?key=${API_KEY}" \
  > synthesize-text.txt

4. nano tts_decode.py

5. python tts_decode.py --input "synthesize-text.txt" --output "synthesize-text-audio.mp3"
```

## Task 3. Perform speech to text transcription with the Cloud Speech API

```python
1. nano request.json
{
  "config": {
      "encoding":"FLAC",
      "languageCode": "fr"
  },
  "audio": {
      "uri":"gs://cloud-samples-data/speech/corbeau_renard.flac"
  }
}


2. curl -s -X POST -H "Content-Type: application/json" \
--data-binary @request.json \
"https://speech.googleapis.com/v1/speech:recognize?key=${API_KEY}" \
> speech_response_fr.json

3. cat speech_response_fr.json
```

## Task 4. Translate text with the Cloud Translation API

```python
TEXT="これは日本語です。"

1. curl -X POST "https://translation.googleapis.com/language/translate/v2?target=en&key=${API_KEY}" \
-d "q=${TEXT}" \
> translation_response.txt

2. cat translation_response.txt
```

## Task 5. Detect a language with the Cloud Translation API

```python
TEXT="Este%é%japonês"

1. curl -X POST "https://translation.googleapis.com/language/translate/v2/detect?key=${API_KEY}" \
-d "q=${TEXT}" \
> detected_response.txt

2. cat detected_response.txt
```
